```python
from deltalake import DeltaTable
import pyarrow as pa
dt=DeltaTable('c:\\Users\\Utkarsh\\Desktop\\Azure DataBricks Project\\Sales-Analytics-Using-Azure-Databricks\\data\\delta\\sales_cleaned')

reader = dt.scan()

table = reader.read_all()

# Convert arro3 Table → PyArrow Table
arrow_table = pa.table(table)

# Convert PyArrow Table → Pandas DataFrame
df_sales = arrow_table.to_pandas()

print(df_sales.head())
```

```python
df_sales['Outlet_Size'] = df_sales['Outlet_Size'].fillna("Unknown")

df_sales['Outlet_Size'].isna().sum()
```

```python
import pandas as pd

df_sales['MRP_Category']= pd.qcut(
df_sales['Item_MRP'], q=4, labels=['Low', 'Medium', 'High', 'Premium']
)

```
