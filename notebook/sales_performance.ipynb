## Data Reading

```python
import pandas as pd

df=pd.read_parquet('c:\\Users\\Utkarsh\\Desktop\\Azure DataBricks Project\\Sales-Analytics-Using-Azure-Databricks\\data\\processed\\sales_cleaned.parquet')
```

```python
import duckdb

duckdb.sql("Select * from df limit 5").df()
```

```python
duckdb.sql("""Select 
           
    Outlet_Identifier,
    Outlet_Type,
    Outlet_Size,
    Outlet_Location_Type,
    Outlet_Establishment_Year,
    SUM(Item_Outlet_Sales) AS Total_Sales,
    AVG(Item_Outlet_Sales) AS Average_Sales

    from df
        
    Group by Outlet_Identifier,
        Outlet_Type,
        Outlet_Size,
        Outlet_Location_Type,
        Outlet_Establishment_Year

order by Average_Sales desc;
    
            """).df()
```

Observation:

Supermarket outlets are associated with substantially higher average sales than Grocery Stores.

Supermarket Type3 is the highest-performing outlet type.

High-performing outlets are not restricted to a single location tier.


Within each outlet, which Item Types contribute the most total sales?

```python
duckdb.sql("""Select 
           
            Outlet_Identifier,
            Item_Type,
            count (*) as Records,
            sum (Item_Outlet_Sales) as Total_Sales,
            avg(Item_Outlet_Sales) as Average_Sales


    from df
        
    Group by Outlet_Identifier,
        Item_Type,
        

order by Outlet_Identifier,
Total_Sales desc;
    
            """).df()
```

OUT010's relatively high total sales within individual product categories are partly explained by having many records, while its low average sales indicate weaker sales performance per product-outlet observation.

```python
duckdb.sql("""Select 
           
            Outlet_Identifier,
            Item_Type,
            count (*) as Records,
            sum (Item_Outlet_Sales) as Total_Sales,
            avg(Item_Outlet_Sales) as Average_Sales


    from df
    where Outlet_Identifier in ('OUT027','OUT010','OUT019')
    Group by Outlet_Identifier,
        Item_Type,
        

order by Outlet_Identifier,
Total_Sales desc;
    
            """).df()
```

Observation:

OUT027 sales are substantially higher across almost all Item Types compared with the two low performing outlets.

OUT027, the only Supermarket Type3 outlet that has the highest average sales.
